```python
import mne

coords, faces = mne.read_surface(r"/data/Schoffelen/derivatives/smriprep/freesurfer/sub-A2002/surf/rh.orig")
mne.write_surface("test.obj", coords, faces, overwrite=True)
```

```python
source_space = r"/data/Valentina Mancini/meeg_forward_workflow/_subject_S01/setup_volume_source_space/S01_src-vol-src.fif"
src = mne.read_source_spaces(source_space, verbose=False)
```

```python
%matplotlib inline
src.plot(subjects_dir='/data/Valentina Mancini/freesurfer')
```

```python
report = mne.Report()

orientations = ['coronal', 'axial', 'sagittal']
for orietation in orientations:
    fig = mne.viz.plot_bem(src=src, subject='S01', subjects_dir='/data/Valentina Mancini/freesurfer', orientation=orietation, brain_surfaces='white', show=False)
    report.add_figure(
        fig=fig,
        title=f"Source spaces ({orietation} view)",
        image_format="PNG",
        section="Source spaces",
        tags=("sources", "bem")
    )
report.save("/data/tmp/report_custom_figure.html", overwrite=True)
```
