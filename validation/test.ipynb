```python
fname = fname_LUT
dtype = [
    ("id", "<i8"),
    ("name", "U"),
    ("R", "<i8"),
    ("G", "<i8"),
    ("B", "<i8"),
    ("A", "<i8"),
]
lut = {d[0]: list() for d in dtype}
with open(fname) as fid:
    for line in fid:
        line = line.strip()
        if line.startswith("#") or not line:
            continue
        line = line.split()
        if len(line) != len(dtype):
            print(line)
            raise RuntimeError(f"LUT is improperly formatted: {fname}")
        for d, part in zip(dtype, line):
            lut[d[0]].append(part)
lut = {d[0]: np.array(lut[d[0]], dtype=d[1]) for d in dtype}
assert len(lut["name"]) > 0
lut["name"] = [str(name) for name in lut["name"]]
```

```python
import nibabel as nib
import pandas as pd

lh = nib.load(r"C:\Users\ferat\Documents\GitHub\meeg-pipelines\meegpype\data\fsLR\atlas_subparc374.L.32k_fs_LR.label.gii")

color_mapping_lh = {}
for label in lh.labeltable.labels:
    color_mapping_lh[label.key] = (label.red, label.green, label.blue, label.alpha)
color_mapping_lh

dict_lh = lh.labeltable.get_labels_as_dict()

data = []
for key in dict_lh.keys():
    label_name = dict_lh[key]
    label_value = key
    label_color = color_mapping_lh[key]
    d = {'label_name': label_name,
            'label_value': label_value,
            'R': int(label_color[0]*255),
            'G': int(label_color[1]*255),
            'B': int(label_color[2]*255),
            'A': int(label_color[3]*255)}
    data.append(d)
df_l = pd.DataFrame(data)

rh = nib.load(r"C:\Users\ferat\Documents\GitHub\meeg-pipelines\meegpype\data\fsLR\atlas_subparc374.R.32k_fs_LR.label.gii")
color_mapping_rh = {}
for label in rh.labeltable.labels:
    color_mapping_rh[label.key] = (label.red, label.green, label.blue, label.alpha)
color_mapping_rh

dict_rh = rh.labeltable.get_labels_as_dict()

data = []
for key in dict_rh.keys():
    label_name = dict_rh[key]
    label_value = key
    label_color = color_mapping_rh[key]
    d = {'label_name': label_name,
            'label_value': label_value,
            'R': int(label_color[0]*255),
            'G': int(label_color[1]*255),
            'B': int(label_color[2]*255),
            'A': int(label_color[3]*255)}
    data.append(d)
df_r = pd.DataFrame(data)

df_l['label_value'] += 1000
df_r['label_value'] += 2000

df = pd.concat([df_l, df_r], ignore_index=True)

# Export as FreeSurfer LUT format
def export_freesurfer_lut(df, output_path):
    """Export DataFrame to FreeSurfer ColorLUT format."""
    with open(output_path, 'w') as f:
        # Write header
        f.write("#$Id: CustomColorLUT.txt\n\n")
        f.write("#No. Label Name:                            R   G   B   A\n\n")
        
        # Write each label
        for _, row in df.iterrows():
            # Format: number, name (padded to ~40 chars), R, G, B, A
            label_num = int(row['label_value'])
            label_name = row['label_name']
            r, g, b, a = int(row['R']), int(row['G']), int(row['B']), int(row['A'])
            
            # Pad label name to align columns (similar to FreeSurfer format)
            f.write(f"{label_num:<20}{label_name:<40}{r:<4}{g:<4}{b:<4}{a}\n")
    
    print(f"LUT saved to: {output_path}")

# Export the LUT
output_lut_path = r"C:\Users\ferat\Documents\GitHub\meeg-pipelines\meegpype\data\fsLR\atlas_subparc374_ColorLUT.txt"
export_freesurfer_lut(df, output_lut_path)

df
```

```python
import os
import matplotlib.pyplot as plt
import numpy as np
import mne
from mne.minimum_norm import make_inverse_operator, source_induced_power



########################################################################################################
# step3

## Load all runs
raws = [r"d:\VM\derivatives\test\sub-07\ses-01\meg\sub-07_ses-01_task-hippo_run-01_proc-tsss_meg.fif", r"d:\VM\derivatives\test\sub-07\ses-01\meg\sub-07_ses-01_task-hippo_run-02_proc-tsss_meg.fif", r"d:\VM\derivatives\test\sub-07\ses-01\meg\sub-07_ses-01_task-hippo_run-03_proc-tsss_meg.fif"]
raw = mne.concatenate_raws([mne.io.read_raw_fif(f, verbose=False) for f in raws])

# load forward solution
fwd_fname = r"D:\VM\derivatives\test\sub-07\ses-01\meg\forward.fif"
fwd = mne.read_forward_solution(fwd_fname)

# Epochs parameters, can be adjusted
tmin, tmax = -0.1, 0.5
baseline = (-0.1, 0.0)
events = mne.find_events(raw)

# define the name of your triggers (need to be adjusted)
event_id = {"stimulus_A": 256,
            "stimulus_B": 512,
            "stimulus_C": 1024}

epochs = mne.Epochs(
    raw, events, event_id=event_id, tmin=tmin, tmax=tmax,
    baseline=baseline, preload=True, reject_by_annotation=True
)

## select only the events you want to study, need to be adjusted
epochs = epochs[["stimulus_A", "stimulus_B"]]

# Noise covariance
cov = mne.compute_covariance(epochs, tmin=baseline[0], tmax=baseline[1], method="empirical", rank=None)

# Inverse operator
info = raw.info
depth = 0.8
fixed = False
loose = {'volume': 1.0}

inverse_operator = make_inverse_operator(info, fwd, cov, loose=loose, depth=depth, fixed=fixed)

########################################################################################################
# step4

aseg_fname = r"D:\VM\derivatives\test\sourcedata\freesurfer\sub-07\mri\aparc+aseg.mgz"
results_dir = "."
freqs = np.arange(7, 101, 2)
n_cycles = freqs / 3.0
n_jobs = 8

# apply inverse operator on epochs
epochs_total = epochs.copy()
epochs_induced = epochs.copy().subtract_evoked()

# -------------------------
# Hippocampal labels
# -------------------------
label_names, _ = mne.get_volume_labels_from_aseg(aseg_fname, return_colors=True)
hipp_labels = [name for name in label_names if "hippocampus" in name.lower()]
left_hipp = [lab for lab in hipp_labels if "left" in lab.lower() or "lh" in lab.lower()]
right_hipp = [lab for lab in hipp_labels if "right" in lab.lower() or "rh" in lab.lower()]

# -------------------------
# Compute power and ITC separately
# -------------------------
def save_label_power_itc(label_name, epochs_total, epochs_induced, inv_op, out_dir):
    for ep, tag in zip([epochs_total, epochs_induced], ["total", "induced"]):
        power, itc = source_induced_power(
            ep, inv_op, freqs=freqs, baseline=baseline,
            baseline_mode="percent", n_cycles=n_cycles, n_jobs=n_jobs
        )

        power_mean = np.mean(power, axis=0)
        itc_mean = np.mean(itc, axis=0)
        times = ep.times

        # Save numpy matrices
        np.save(os.path.join(out_dir, f"{label_name}_power_{tag}.npy"), power_mean)
        np.save(os.path.join(out_dir, f"{label_name}_itc_{tag}.npy"), itc_mean)

        # Power figure
        fig_p, ax_p = plt.subplots(figsize=(12, 4), constrained_layout=True)
        vmax_p = np.max(np.abs(20 * power_mean)) if power_mean.size else 1.0
        im_p = ax_p.imshow(
            20 * power_mean,
            extent=[times[0], times[-1], freqs[0], freqs[-1]],
            aspect="auto", origin="lower",
            cmap="RdBu_r", vmin=-vmax_p, vmax=vmax_p
        )
        ax_p.set_title(f"{label_name} - Power ({tag})")
        fig_p.colorbar(im_p, ax=ax_p)
        fig_p.savefig(os.path.join(out_dir, f"{label_name}_power_{tag}.png"))
        plt.close(fig_p)

        # ITC figure
        fig_itc, ax_itc = plt.subplots(figsize=(12, 4), constrained_layout=True)
        itc_centered = itc_mean - np.mean(itc_mean) if itc_mean.size else itc_mean
        vmax_itc = np.max(np.abs(itc_centered)) if itc_centered.size else 1.0
        im_itc = ax_itc.imshow(
            itc_centered,
            extent=[times[0], times[-1], freqs[0], freqs[-1]],
            aspect="auto", origin="lower",
            cmap="PiYG", vmin=-vmax_itc, vmax=vmax_itc
        )
        ax_itc.set_title(f"{label_name} - ITC ({tag})")
        fig_itc.colorbar(im_itc, ax=ax_itc)
        fig_itc.savefig(os.path.join(out_dir, f"{label_name}_itc_{tag}.png"))
        plt.close(fig_itc)

        print(f"[INFO] Saved {tag} power and ITC for {label_name}")

# -------------------------
# Run for all hippocampal labels
# -------------------------
for lab in left_hipp + right_hipp:
    save_label_power_itc(lab, epochs_total, epochs_induced, inverse_operator, results_dir)



```
